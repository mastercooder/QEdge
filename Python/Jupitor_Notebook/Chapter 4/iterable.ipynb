<h2><b><u>4. Yield</u></b></h2>
<i>

````python
def f1():
    for i in range(1, 11):
        print(i, end=" ")
````

````python
f1()
````

````python
def f2():
    for i in range(1, 11):
        return i
````

````python
f2()
````

````python
def f3():
    print('hello')
    return "Okai"
    print("thanks")
````

````python
f3()
````

````python
def f4():
    for i in range(1, 11):
        yield i
````

````python
f4()
````

````python
gen2 = f4()
````

````python
type(gen2)
````

````python
next(gen2)
````

````python
gen2.__next__()
````

````python
for i in gen2:
    print(i, end=" ")
````

````python
def f5():
    yield 100
    yield 200
    yield 300
````

````python
f5()
````

````python
gen3 = f5()
````

````python
type(gen3)
````

````python
next(gen3)
````

````python
gen3.__next__()
````

````python
gen3.__next__()
````

````python
gen3.__next__()
````
