````python
def table(n):
    def ntable(x):
        print("*********************")
        for i in range(1, x+1):
            print(n, "*", i, "=", n*i)
        
        print("*********************")
    return ntable
````

````python
table5 = table(5)
````

````python
table5(10)
````

````python
table4 = table(4)
````

````python
table4(4)
````

````python
def discount(d):
    def sales(p):
        return p-(p/100*d)
    return sales
````

````python
daily_sales = discount(10)
````

````python
daily_sales(1000)
````

````python
weekly_sales = discount(20)
weekly_sales(1000)
````

<h2><b> 2. <u> Decorator in Python </u></b></h2>

<i>
Suppose when i run login1() and i need output as shown below then Decorator in python is used 
</i>

```json
  Output
----------


****************check before***************
Logout done
*****************maintain after*************
```



````python
def f1(fun):
    def f2():
        print("****************check before***************")
        fun()
        print("*****************maintain after*************")
    return f2
````

````python
def login():
    print("Login Success")
````

````python
login()
````

<h3><i>Here @ is Decorator</i></h3>
<h3><i>After @ function must be clouse function</i></h3>

````python
@f1
def login():
    print("Login Successfull")
````

````python
login()
````

````python
@f1
def logout():
    print("Logout Done")
````

````python
logout()
````
