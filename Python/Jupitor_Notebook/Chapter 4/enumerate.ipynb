<h1><b><u>Enumerate</u></b></h1>

<ol>
<li>Enumerate is a in-build class</li>
<li>It is used to create individual sequence of numbers for each element of iterable objects  </li>
<li>Enumerate object is iterator, so we can read data only one time</li>
<li>syntax --> enumerate(iterableObject, initial=0)</li>
</ol>

<h2><b><u>1. Enumerate</u></b></h2>

```python
type(enumerate)
```

```python
courses = ['python', 'java', 'react', 'ML', 'DL', 'NLP', 'Gen-AI']

for i in courses:
    print(i)
```

```python
for i in enumerate(courses):
    print(i)
```

```python
for i in enumerate(courses, 1001):
    print(i)
```

```python
for cid, cname in enumerate(courses, 1001):
    print(cid, cname)
```

```python
enum1 = enumerate(2345252352334, 101)
```

```python
enum1 = enumerate(str(2345252352334), 1001)

for i in enum1:
    print(i)
```

```python
enum3 = enumerate({1: 100, 2: 200, 3: 300, 4: 400}, 9001)

for i in enum3:
    print(i)
```

```python
enum4 = enumerate({1: 100, 2: 200, 3: 300, 4: 400}.values(), 9001)

for i in enum4:
    print(i)
```

```python
enum5 = enumerate({1: 100, 2: 200, 3: 300, 4: 400}.items(), 9001)

for i in enum5:
    print(i)
```

<h2><b><u>2. Itertools</u></b></h2>

<i> count can be used to start couting from there 

```python
from itertools import count 
```

```python
batchId = count(507)
```

```python
enum6 = enumerate(['hari', 'giri', 'suri'], next(batchId))
```

```python
for i in enum6:
    print(i)
```

```python

batchID2 = count(1)

enum7 = enumerate({1: 100, 2: 200, 3: 300, 4: 400, 5: 500, 6: 600, 7: 700}.items(), next(batchID2))

for i in enum7:
    print(i)
```
