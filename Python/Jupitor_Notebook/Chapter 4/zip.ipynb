<h1><b><u>Zip</u></b></h1>

1. Zip is python in-build class
2. Zip will read two more iterable objects element parallel
3. it paris up elements based on their posistions 
4. All first element together and second element together ....

```python
type(zip)
```

```python
courses = ['python', 'react', 'java', 'next', 'django', 'fastapi', 'docker']
```

```python
fees = [3500, 2000, 4000, 2500, 4400, 3200, 1500]
```

python 3500
react 2000

```python
for course in courses:
    print(course)
```

```python
for fee in fees:
    print(fee)
```

```python
for course, fee in courses, fees:
    print(course, fee)
```

```python
for i in zip(courses, fees):
    print(i)
```

```python
for course, fee in zip(courses, fees):
    print(course, fee)
```

```python
cids = [101, 102, 103, 104, 105, 106, 107]
```

```python
for cid, course, fee in zip(cids, courses, fees):
    print(cid, course, fee)
```

```python
for cousre, fee in enumerate(zip(courses, fees), 7001):
    print(cousre ,fee)
```

```python
for cousre, fee in enumerate(zip(courses, fees), 7001):
    print(cousre , *fee)
```

```python
for cousre, fee in enumerate(zip(courses, fees), 7001):
    print("Course: ", cousre)
    print("fees: ", *fee)
```
