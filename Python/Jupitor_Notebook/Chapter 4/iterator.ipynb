<h1><b><u>Iterator</u></b></h1>

<ol>
<li>Iterator is a function</li>
<li>We can create iterator object from existing iterable objects with the help of iter()</li>
<li>We can read data in forward direction only from iterator that means we can read data only one time from iterator</li>
</ol>


```python
type(iter)
```

<h4><i>We can not create empty iterator</i></h4>

```python
itr1 = iter()
```

<h4><i>iterator syntax ---> iter1 = iter(iterableobjectOnly)</i></h4>
<h4><i>iterable object ---> string, range, list, tuple, dictionary, set, dict.keys, dict.values, dict.items</i></h4>

```python
iter1 = iter("welcome")
```

<h2><b><u>1. Reading data from iterator</u></b></h2>

```python
iter1[0]
```

```python
iter1[:]
```

```python
print(iter1)
```

```python
for i in iter1:
    print(i)
```
