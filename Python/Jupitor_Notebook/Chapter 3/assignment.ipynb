6. Write a funtion to find given number is prime or not
    funtion name ==> is_prime(7)

```python
def is_prime(user):
    for i in range(2, user):
        if(user % i == 0):
            return "{} is Not a Prime".format(user)
            break
        
        else:
            return "{} is a Prime Number".format(user)
            break
 

try:
    user = int(input("Enter any Number: "))
    result = is_prime(user)
    print(result)
except Exception:
    print("Enter a Valid Input")
```

7. Write a recusive funtion for Fibonacci series
    function name ==> fib(45)

```python
def fib(num):
    a, b = 0, 1
    for i in range(num):
        print(a, end=" ")
        a, b = b, a+b

num = 45
fib(num)
```

8. Find mobile number 
    write a function to find 10 digit number from given text

```python
import re

def noFinder(text):
    pattern = r'\b[6-9]\d{9}\b'
    return re.findall(pattern, text)

text = "My number is 9876543210 and my friend's number is 9123456780. Invalid: 12345"

result = noFinder(text)
print(result)
```
