
<h2><b><u>3. Creating Table And Inserting Records</u></b></b2></h2>

<h3><b> i.  <u>Making Connection</u>

```python
import psycopg2

conn = psycopg2.connect(
    dbname='postgres', 
    user='postgres',
    password='14422',
    host='localhost',
    port='5432'
    )


cur = conn.cursor()
```

<h3><b> ii. <u> Createing Table

```python
cur.execute("""
CREATE TABLE lerner(
    lid INTEGER PRIMARY KEY,
    lname TEXT,
    email TEXT,
    city TEXT
)       
""")

conn.commit()

print("Table Created successfully")
```

<h3><b> iii. <u> Inserting one Record in the Table

```python
cur.execute(
    "INSERT INTO lerner values(7002, 'ramu', 'ramu@zohomail.in','channi')"
)
conn.commit()

print("Data Inserted Successfully")
```

<h3><b> iv. <u> Inserting Multiple Records in The Table

```python
lernerData = [
    (7002, 'venkat', 'venkat@zohomail.in', 'Delhi '),
    (7003, 'ramesh', 'ramesh@gmail.com', 'Mumbai'),
    (7004, 'suresh', 'suresh@yahoo.com', 'Hyderabad'),
    (7005, 'priya', 'priya@outlook.com', 'Bengaluru'),
    (7006, 'anita', 'anita@hotmail.com', 'Pune')
]
```

```python
lernerData
```

```python
import psycopg2

conn = psycopg2.connect(
    dbname='postgres', 
    user='postgres',
    password='14422',
    host='localhost',
    port='5432'
    )


cur = conn.cursor()

cur.executemany(
    "INSERT INTO lerner(lid, lname, email, city) values(%s,%s,%s,%s)", 
    lernerData
)

conn.commit()

print("Leraner Data Inserted Successfully!")
```


<h2><b><u>4. Deleting Data From Table</u></b></b2></h2>

<h3><b> i.  <u>Making Connection</u>

```python
import psycopg2

cunn = psycopg2.connect(
    dbname = 'postgres',
    user = 'postgres',
    password = '14422',
    host = 'localhost'
)

cur = cunn.cursor()
```

<h3><b> ii.  <u>Deleting Operation</u>

```python
lid = int(input("Enter LID: "))

cur.execute("DELETE FROM lerner WHERE lid={}".format(lid))

cunn.commit()

print(f"{cur.rowcount} Rows Deleted Successfully")
```

<h3><b> iii.  <u>Closing Connection</u>

```python
cur.close()
conn.close()
```


<h2><b><u>5. Update Records in Table</u></b></b2></h2>

<h3><b> i.  <u>Making Connection</u>

```python
import psycopg2

conn = psycopg2.connect(
    dbname = 'postgres',
    user = 'postgres',
    password = '14422',
    host = 'localhost'
)

cur = conn.cursor()
```

<h3><b> ii.  <u>Updating Logic</u>

```python
lid = int(input("Enter the LID: "))
city = input("Enter the City Name: ")

cur.execute("UPDATE lerner SET city='{}' WHERE lid = {}".format(city, lid))

conn.commit()

print(f"{cur.rowcount} Rows Effected")
```

<h3><b> iii.  <u>Closing Connection</u>

```python
cur.close()
conn.close()
```
