<h1><b><u> MongoDB </u></b><h1>

<h3><b> What is MongoDB? </b></h3>

**MongoDB** is a **document-oriented NoSQL database** where data is stored as **documents** (JSON-like objects) inside **collections**.

Unlike traditional relational databases, MongoDB stores data in a flexible format without requiring a fixed schema.

---

<h3><b> Traditional Database (RDBMS) vs MongoDB </b></h3>

| Traditional Database (RDBMS) | MongoDB                             |
| ---------------------------- | ----------------------------------- |
| Tables                       | Collections                         |
| Rows                         | Documents (JSON Objects)            |
| Fixed Schema                 | Flexible (Schema-less)              |
| Structured Data              | Unstructured / Semi-Structured Data |
| Vertical Scaling             | Horizontal Scaling                  |

---

<h3><b> SQL Table Example (Traditional Database) </b></h3>

<h4><u> Student Table </u></h4>

| SID  | SNAME  | COURSE | FEE   |

**Observation:**
Each document can have different fields. MongoDB does not require all documents to have the same structure.

---

<h2><b><u> MongoDB Terminology </u></b></h2>

| MongoDB Term | Description                             |
| ------------ | --------------------------------------- |
| Database     | Container for Collections               |
| Collection   | Group of Documents (similar to a Table) |
| Document     | JSON-like Record (similar to a Row)     |
| Field        | Key-Value Pair inside a Document        |

---

<h2><b><u> Why Do We Use MongoDB? </u></b></h2>

<h4><b> 1. Flexible Schema </b></h4>

* Easy to modify the data structure.
* New fields can be added without affecting existing documents.

<h4><b> 2. Stores Complex Data Easily </b></h4> 

* Supports nested objects and arrays.
* Ideal for hierarchical data.

<h4><b> 3. High Performance </b></h4> 

* Fast read and write operations.
* Suitable for handling large amounts of data.

<h4><b> 4. Horizontal Scaling </b></h4> 

* Data can be distributed across multiple servers.
* Supports large-scale applications.

---

<h2><b><u> Applications of MongoDB </u></b></h2>

MongoDB is widely used in:

1. Web Application Development
2. Real-Time Analytics
3. Content Management Systems (CMS)
4. E-Commerce Platforms
5. Mobile Applications
6. Big Data Applications

---

<h2><b><u> Conclusion </h2></b></u>

**MongoDB** is a flexible, JSON-based NoSQL database designed for modern applications. It provides high performance, scalability, and schema flexibility, making it suitable for handling large volumes of structured and unstructured data.
