```python
# Exceptions
# try....except....else....finally...
# try---> Contains Error code
# exception --> will execute when there is error existed
# else ---> if no erro then this block will execute
# finally --> this block will execute always
# case 1 ---> Error riased 
# case 2 ---> No error
# case 3 ---> errr is there but not mentioned in except block
```

```python
#  Exception Syntax 
# ===================
# try:
#     error code
# except (er1, er2, er3..., ern):
#     code on no error
# finally: 
```

```python
v1 = input("Enter a integer Value for V1: ")
v2 = input("Enter a integer Value for V2: ")

try:
    v3 = int(v1) + int(v2)
except Exception as e:
    print("Error: ", e.with_traceback)
else:
    print(v3)
finally: print("Thanks")
```

```python
v1 = input("Enter a integer Value for V1: ")
v2 = input("Enter a integer Value for V2: ")

try:
    v3 = int(v1) + int(v2)
except Exception as e:
    print("Error Message: ", e.args[0])
else:
    print(v3)
finally: print("Thanks")
```

```python
key = input("Enter Key to get Value: ")
try:
    if(key.isdigit()):
        value = D1[int(key)]
    else:
        value = D1[key]
except(KeyError):
    print("Given Key Not Existed")
else:
    print(value)
finally:
    print("Thanks")
```

```python
courses = {'python': 12000, 'react': 8000, 'django': 7000}
```

```python
course = input("Enter you course Name: ")
if course in courses:
    fee = courses.get(course)
    print("Fee for you given coures {} is {}".format(course, fee))
else:
    print("Given Course is not Found")
```

```python
courses = {'python': 12000, 'react': 8000, 'django': 7000}

course = input("Enter you course Name: ")
try:
    if course in courses:
        fee = courses.get(course)
    else:
        raise CourseNotFoundError
except(CourseNotFoundError):
    print("Given Course is not Existed")
else:
    print("Fee for you given coures {} is {}".format(course, fee))
finally:
    print("Thanks for Visiting")
```
