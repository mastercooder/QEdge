# Regular Expression

## 1. Anchors (Position-based)

| Symbol  | Name          | Meaning                 | Example    | Matches        |
| ------- | ------------- | ----------------------- | ---------- | -------------- |
| `^`     | Start anchor  | Start of string         | `^abc`     | `"abc123"` ✔   |
| `$`     | End anchor    | End of string           | `abc$`     | `"123abc"` ✔   |
| `^...$` | Full match    | Whole string must match | `^[0-9]+$` | `"123"` ✔      |
| `\b`    | Word boundary | Start/end of word       | `\bcat\b`  | `"cat"` ✔      |
| `\B`    | Not boundary  | Inside word             | `\Bcat\B`  | `"scatters"` ✔ |


## 2. Character Classes

| Pattern  | Meaning           | Example Match |
| -------- | ----------------- | ------------- |
| `[abc]`  | a or b or c       | `"a"`         |
| `[a-z]`  | lowercase letters | `"m"`         |
| `[A-Z]`  | uppercase letters | `"T"`         |
| `[0-9]`  | digits            | `"7"`         |
| `[^abc]` | NOT a, b, c       | `"x"`         |


## 3. Predefined Classes (Shortcuts)

| Pattern | Meaning        | Equivalent      |
| ------- | -------------- | --------------- |
| `\d`    | digit          | `[0-9]`         |
| `\D`    | NOT digit      | `[^0-9]`        |
| `\w`    | word char      | `[A-Za-z0-9_]`  |
| `\W`    | NOT word       | `[^A-Za-z0-9_]` |
| `\s`    | whitespace     | space, tab      |
| `\S`    | NOT whitespace | no spaces       |


## 4. Quantifiers (Repetition)

| Symbol  | Meaning   | Example   | Matches        |
| ------- | --------- | --------- | -------------- |
| `+`     | 1 or more | `a+`      | `"a", "aa"`    |
| `*`     | 0 or more | `a*`      | `"" , "aaa"`   |
| `?`     | 0 or 1    | `a?`      | `"" , "a"`     |
| `{n}`   | exactly n | `\d{3}`   | `"123"`        |
| `{n,}`  | n or more | `\d{2,}`  | `"12", "123"`  |
| `{n,m}` | range     | `\d{2,4}` | `"12", "1234"` |


## 5. Special Symbols

| Symbol | Meaning                   | Example          |      |      |
| ------ | ------------------------- | ---------------- | ---- | ---- |
| `.`    | any char (except newline) | `a.b`            |      |      |
| `      | `                         | OR               | `cat | dog` |
| `()`   | grouping                  | `(ab)+`          |      |      |
| `\`    | escape                    | `\.` matches `.` |      |      |


## 6. Lookaheads (Conditions)

| Pattern   | Meaning                               |
| --------- | ------------------------------------- |
| `(?=...)` | must contain (positive lookahead)     |
| `(?!...)` | must NOT contain (negative lookahead) |


## 7. REGEX FUNCTIONS

| Function         | Syntax               | What it Does                | Where it Checks | Returns         | Example                     | Output          |
| ---------------- | -------------------- | --------------------------- | --------------- | --------------- | --------------------------- | --------------- |
| `re.match()`     | `re.match(p, s)`     | Match pattern at start      | Beginning only  | Match / None    | `re.match("a","abc")`       | ✔               |
| `re.search()`    | `re.search(p, s)`    | Find first match            | Anywhere        | Match / None    | `re.search("b","abc")`      | ✔               |
| `re.fullmatch()` | `re.fullmatch(p, s)` | Match entire string         | Full string     | Match / None    | `re.fullmatch("\d+","123")` | ✔               |
| `re.findall()`   | `re.findall(p, s)`   | Find all matches            | Anywhere        | List            | `re.findall("\d+","a1b22")` | `['1','22']`    |
| `re.finditer()`  | `re.finditer(p, s)`  | Find all matches (detailed) | Anywhere        | Iterator        | loop over matches           | Match objects   |
| `re.sub()`       | `re.sub(p, r, s)`    | Replace matches             | Anywhere        | New string      | `re.sub("\d","X","a1")`     | `"aX"`          |
| `re.split()`     | `re.split(p, s)`     | Split string                | Anywhere        | List            | `re.split("[,;]","a,b;c")`  | `['a','b','c']` |
| `re.compile()`   | `re.compile(p)`      | Precompile pattern          | —               | Pattern object  | `pat = re.compile("\d+")`   | reusable        |
| `re.subn()`      | `re.subn(p, r, s)`   | Replace + count             | Anywhere        | (string, count) | `re.subn("\d","X","a1")`    | `("aX",1)`      |
| `re.escape()`    | `re.escape(s)`       | Escape special chars        | —               | String          | `re.escape(".+")`           | `\.\+`          |

<h1><b> Group

```python
# Us Contact Number ==> 567-456-5823
contact = '567-456-5823'
```

```python
contactGrp1 = re.search("([1-9]{3})-([0-9]{3})-([0-9]{4})", contact)
```

```python
contactGrp1.groupdict()
```

```python
contactGrp1.groups()
```

```python
print(contactGrp1.group(0))
print(contactGrp1.group(1))
print(contactGrp1.group(2))
print(contactGrp1.group(3))
```

```python
for i in contactGrp1.groups():
    print(i)
```

```python
contactGrp2 = re.search("(?P<State>[1-9]{3})-(?P<Exchange>[0-9]{3})-(?P<Phone>[0-9]{4})", contact)
```

```python
contactGrp2.groupdict()
```

```python
contactGrp2.groupdict()['State']
contactGrp2.groupdict()['Exchange']
contactGrp2.groupdict()['Phone']
```
