```python
# re is a module to implement python regular expressions 
import re
```

```python
# match
re.match
```

```python
# what is pattern?
# pattern is that which you called expecting
# string  ---> where you want to check you pattern
```

```python
# syntax of match funtion 
re.match(pattern, string)
re.match ---> return 'match object' if it matches pattern in string ===> span, string
re.match ---> return 'None' if it doesn't matches pattern in string ===> None
```

```python
print(re.match("python", "python is good for AI application"))
print(re.match("python", "xpython is good for AI application"))
```

```python
print(re.match("python", "python is good for AI application"))
print(re.match("python", "i have joined python fullstack cource"))
print(re.match("PYTHON", "python is good for AI applications"))
print(re.match("PYTHON", "python is good for AI applications", re.I))
print(re.match("PYTHON", "python is good for AI applications", re.IGNORECASE))
```

```python
# meta characters
# disign pattern with meta character
# []  ^ {m, n} + (option1|option2|option3) . ? *
```

```python
# [0123456789]  --> it matches at lest one digit 
str1 = input("Enter a string value: ") # 7 a @ 
if re.search('[0-9]', str1):
    print("Matched")
else:
    print("Not Matched")
```

```python
# [0-9][0-9]  --> it matches at lest two digit 
str1 = input("Enter a string value: ") # 7 a @ 
if re.search('[0-9][0-9]', str1):
    print("Matched")
else:
    print("Not Matched")
```

```python
# ^ caret || Tell program to match the first character
str1 = input("Enter a string value: ") # 7 a @ 
if re.search('^[0-9]', str1):
    print("Matched    ")
else:
    print("Not Matched")
```

```python
# $ dollar | Matches the last character
str1 = input("Enter a string value: ") # 7 a @ 
if re.search('[0-9]$', str1):
    print("Matched    ")
else:
    print("Not Matched")
```

```python
# ^ and $ || only one ditit number
str1 = input("Enter a string value: ") # 7 a @ 
if re.search('^[0-9]$', str1):
    print("Matched    ")
else:
    print("Not Matched")
```

```python
# ^ and $ || only two ditit number
str1 = input("Enter a string value: ") # 7 a @ 
if re.search('^[0-9][0-9]$', str1):
    print("Matched    ")
else:
    print("Not Matched")
```
