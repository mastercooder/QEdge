```python
# fileobject status related methods
```

```python
fileName = r"D:"
```

```python
fobj = open(fileName, 'r')
```

```python
print(fobj.name)
print(fobj.mode)
print(fobj.fileno())
print(fobj.readable())
print(fobj.writable())
print(fobj.seekable())
print(fobj.close()) # None ---> means files object successfully closed
print(fobj.closed)
```
