<h2><b> <u> Assignment

1. Write a Function 
    1. Ask Username and course
    2. If Course is existed in course then display fee else display no such course is avalable
    3. Ask Fee to pay for given course
    4. It has to generate a new Student ID
    5. Write the data into student.txt file

```python
import re

courses = {
    'python': 12000,
    'react': 4500,      
    'django': 7000,
    'fastapi': 5500,
    'java': 8000,
    'javascript': 6000
}

fileName = "txt files/students"

sid = input("Enter Sudent ID: ").replace(" ", "")

for i in open('txt files/students.txt'):
    if re.fullmatch(re.escape(sid),i.split(",")[0].replace(" ", "")):
        cname = i.split(",")[2].replace(" ", "").lower()
        std_name = i.split(",")[1].replace(" ", "").capitalize()
        std_fee = int(i.split(",")[3].replace(" ", ""))
        course_fee = courses[cname]
        due_fee = course_fee - std_fee
        
        if due_fee == 0:
            print("{} Your Fees is Clear".format(std_name))
            break
else:
    print("{} Your Due fee is {}".format(std_name, due_fee))
```

<h3><b> i. <u>Append Mode of a file which is Existing

```python
fname = 'txt files/students2.txt'

fobj = open(fname, 'a')
fobj.tell()

fobj.write("\n1012, patel, react, 8000")

fobj.close()
```
