<i><h3><b>i. <u>dumps</u></b></h3>
<i>dumps method is going to convert python native object to json object

```python
jsonObject = json.dumps(student)
```

```python
jsonObject
```

```python
type(jsonObject)
```

<i><h3><b>ii. <u>dump</u></b></h3>
<i>dump method will read python native object data and Store into a json File (student.json)

```python
student = {
    'first_name': 'Tejesh', 
    'last_name': 'Patel', 
    'course': 'python', 
    'fee': 12500,
    'emil': 'tejesh@zohomail.in'
}
```

```python
type(student)
```

```python
fobj = open('txt files/student.json', 'w')

json.dump(student, fobj)

fobj.close()
```

<i><h3><b>iii. <u>jsob.loads</u></b></h3>
<i>json object converts into python native object(Dictionary)

```python
jsonStudentData = '''
{
    "first_name": "Tejesh", 
    "last_name": "Patel", 
    "course": "python", 
    "fee": 12500
}
'''
```

```python
type(jsonStudentData)
```

```python
studentInfo = json.loads(jsonStudentData)
```

```python
studentInfo
```

```python
type(studentInfo)
```

<i><h3><b>iv. <u>jsob.load</u></b></h3></i>
<li>Reading json file datat from txt files/student.json convert into python native object(Dictionary).</li>
<li>Load method required input as .json file object which is created with open inbuild funtion.</li></i>

```python
# Here Reading student.json file data and creating jsonFileData object with student.json data
jsonFileData = open('txt files/student.json')
```

```python
studentData = json.load(jsonFileData)
```

```python
print(studentData)
```

```python
type(studentData)
```

```python
jsonFileData.close()
```
