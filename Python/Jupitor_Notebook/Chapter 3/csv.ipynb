<h3><b>ii. <u>Reading csv file that Contains # Delimiter</u></b></h3>

```python
fname = 'txt files/students3.csv'
fobj = open(fname)

csvReadObj = csv.reader(fobj, dialect="Hash")

for i in csvReadObj:
    print(i[0])
    
fobj.close()
```

```python
csv.register_dialect("SemiColumn", delimiter=";", skipinitialspace=True)
```

```python
fname = 'txt files/students4.csv'
fobj = open(fname)

csvReadObj = csv.reader(fobj, dialect="SemiColumn")

for i in csvReadObj:
    print(i[0])
    
fobj.close()
```

<h3><b>i. <u> Write a Function to find the Fee Paid by Student</u></b></h3>
<li>Input    --------->   Student ID</li>
<li>Output   --------->   STudent Fee and Due</li>

<h4><u><b>Steps</b></u>
<ol>
<li>Ask Student ID
<li>Find Course name for SID
<li>Find Actual Fee for Student Course
<li>Find Due fee for Student ID
<li>Display On Console
</ol>

```python
course = {'python': 4500, 'react': 3500, 'django': 6500, 'fastapi': 4400, 'nextjs': 7500}
```

```python
import re

sid = input("Enter Student ID: ")

for data in open('txt files/students.csv', 'r'):
    if re.fullmatch(sid, data.split(",")[0]):
        coursName = data.split(",")[2].replace(" ", "")
        studentName = data.split(",")[1].replace(" ", "").capitalize()
        CourseFee = course[coursName]
        PaidFee = int(data.split(",")[3])
        DueFee = CourseFee - PaidFee
        
print("{} Due Fee is: {}".format(studentName, DueFee))
```
