```python
#Data Types: Numberical (int, float, complex, str)

a = 50
print(type(a))

b = 50.0
print(type(b))

c = 2 + 5j
print(type(c))
```

```python
# Data Types: Sequential DataTypes ( list, )

d = [1, 2, 3]
print(type(d))

e = (1, 2, 3)
print(type(e))

for i in range(2):
    print(i)
print(type(range))

f = "This is a String"
print(type(f))

```
