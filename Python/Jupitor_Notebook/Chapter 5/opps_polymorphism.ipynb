<h2><b>1. <u>Method Over Loading</u></b></h2>
<i><b>Class will have only one method with same name</b></i>

```python
class c1:
    a = 100
    
    def m1(self):
        return 'First m1'
    
    def m1(self):
        return 'Second m1'
    
    def m1(self):
        return 'Third m1'
```

```python
c1obj1 = c1()
```

```python
c1obj1.a
```

```python
c1obj1.m1()
```

```python
class c1:
    a = 100
    
    def m1(self, a):
        return 'First m1', a
    
    def m1(self, a, b):
        return 'Second m1', a, b
    
    def m1(self, a, b, c):
        return 'Third m1', a, b, c
```

```python
c1obj1 = c1()
```

```python
c1obj1.m1(10, 20, 30)
```

```python
from pythonlangutil.overload import Overload, signature
```

```python
class c1:
    name = 'Ravi'
    
    @Overload
    @signature()
    def m1(self):
        return 'First m1'
    
    @m1.overload
    @signature('int')
    def m1(self, a):
        return 'Second m1', a
    
    
    @m1.overload
    @signature('str')
    def m1(self, a):
        return 'Third m1', a
    
    
    @m1.overload
    @signature('int', 'str')
    def m1(self, a, b):
        return 'Forth m1', a, b
    
    
    @m1.overload
    @signature('str', 'int') 
    def m1(self, a, b):
        return 'Fifth m1', a, b
    
    
    @m1.overload
    @signature('int', 'int')
    def m1(self, a, b):
        return 'Sixth m1', a, b
    
    
    @m1.overload
    @signature('str', 'str')
    def m1(self, a, b):
        return 'Seventh m1', a, b
```

```python
c1obj1 = c1()
```

```python
c1obj1.m1()
```

```python
c1obj1.m1(100)
```

```python
c1obj1.m1('welcome')
```

```python
c1obj1.m1(100, 'tejesh')
```

```python
c1obj1.m1('tejesh', 100)
```

```python
c1obj1.m1(10, 20)
```

```python
c1obj1.m1('welcome', 'tejesh')
```

<h2><b>2. <u>Method Overriding</u></b></h2>


```python
class c1:
    a = 100
    
    def m1(self):
        return 'm1 of class c1'
    
class c2(c1):
    b = 200
    def m1(self):
        return 'm1 of class c2'
```

```python
c2obj1 = c2()
```

```python
c2obj1.b
```

```python
c2obj1.m1()
```

```python
class c1:
    a = 100
    
    def m1(self):
        return 'm1 of class c1'
    
class c2(c1):
    b = 200
    def m1(self):
        return 'm1 of class c2', super().m1()
```

```python
c2obj1 = c2()
```

```python
c2obj1.m1()
```

```python
class c1:
    a = 100
    def m1(self):
        return 'Good Morning'
    
class c2(c1):
    b = 200
    def m1(self, name):
        return super().m1(), name
```

```python
c2obj1 = c2()
```

```python
print(c2obj1.m1('tejesh'))
```

```python
class c1:
    a = 100
    def m1(self, sal):
        return sal+(sal*0.4)+(sal*0.3)+(sal*0.2)
    
class c2(c1):
    b = 200
    def m1(self, sal):
        return super().m1(sal)
```

```python
c2obj1 = c2()
```

```python
c2obj1.m1(1000)
```

```python
class c3(c1):
    b = 200
    def m1(self, sal):
        totalSal = super().m1(sal)
        grossSal = totalSal-(totalSal*0.05)
        return grossSal
```

```python
c3obj1 = c3()
```

```python
c3obj1.m1(1000)
```
