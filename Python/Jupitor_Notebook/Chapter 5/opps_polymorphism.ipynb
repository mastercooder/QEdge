<h2><b>i. <u>Method Over Loading</u></b></h2>
<i><b>Class will have only one method with same name</b></i>

```python
class c1:
    a = 100
    
    def m1(self):
        return 'First m1'
    
    def m1(self):
        return 'Second m1'
    
    def m1(self):
        return 'Third m1'
```

```python
c1obj1 = c1()
```

```python
c1obj1.a
```

```python
c1obj1.m1()
```
