<h1 align="center"><b><u>Class and Object in Python</u></b></h1>

<hr>

<h2><b><u>What is a Class?</u></b></h2>

<ul>
    <li>A <b>Class</b> is a blueprint or template used to create objects.</li>
    <li>Just like a house blueprint is used to build multiple houses, a class is used to create multiple objects.</li>
    <li>A class contains <b>Properties</b> and <b>Activities</b>.</li>
    <li>In Python:
        <ul>
            <li><b>Properties</b> are called <b>Attributes</b>.</li>
            <li><b>Activities</b> are called <b>Methods</b>.</li>
        </ul>
    </li>
</ul>

<p><b>A Class contains two components:</b></p>

<ol>
    <li><b>Attributes</b> (Variables)</li>
    <li><b>Methods</b> (Functions)</li>
</ol>

<ul>
    <li>Attributes store data related to an object.</li>
    <li>Methods define the behavior of objects created from the class.</li>
    <li>Every object created from a class has access to its attributes and methods (except private members).</li>
</ul>

<hr>

<h2><b><u>Class Syntax</u></b></h2>

<pre>
class ClassName(object):
    attributes
    methods
</pre>

<hr>

<h2><b><u>Creating a Class</u></b></h2>

```json
class <className>:
    <attr1> = <value1>
    <attr2> = <value2>

    def <methodName>(parameters):
        val1 = value1
        val2 = value2
        statement1
        statement2
        ...
        ...
        return <output>

    def <methodName>(parameters):
        val1 = value1
        val2 = value2
        statement1
        statement2
        ...
        ...
        return <output>
```

<hr>

<h2><b><u>Object Creation Syntax</u></b></h2>

<pre>
objectName = ClassName(arguments)
</pre>

<hr>

<h2><b><u>Example</u></b></h2>

<pre>
class Student:

    college = "Kalinga University"
    course = "BCA AI-ML"

    def display(self):
        print("Student Details")

s1 = Student()
s2 = Student()
</pre>

<p>
Here, <b>s1</b> and <b>s2</b> are objects (instances) of the <b>Student</b> class.
</p>

<h2><b>1. <u>Sample Class Creation and Object Creation</u></b></h2>

````python

class x:
    a = 100
    b = 200
    
    #Object Method if you pass 'self' in the function 
    def m1(self):
        return 'm1 of class x'
    
    def m2(self):
        return 'm2 of class x'
````

````python
x.a
````

````python
x.b
````

````python
x.m1()
x.m2()
````

````python
xObj1 = x()
````

````python
print(xObj1.a)
print(xObj1.b)
````

````python
xObj1.m1()
````

````python
xObj1.m2()
````

````python
class x:
    a = 100
    
    # Not a object method as no 'self' is passed in the function
    def m1():
        return 'm1 of class x'
````

````python
x.m1()
````

````python
xobj1 = x()
````

````python
xobj1.m1()
````

<h2><b>2. <u>Access Class Attribute inside method</u></b></h2>

````python
class x:
    a = 100  # Class Attribute
    
    def m1(self):
        return "m1 of class x", a
````

````python
xobj1 = x()
````

````python
xobj1.a
````

````python
xobj1.m1()
````

````python
class x:
    a = 100
    
    def m1(self):
        a = 120     # variable inside method only(scope)
        return 'm1 of class x', self.a, a
````

````python
xobj1 = x()
````

````python
xobj1.m1()
````

<h3><b>iii. Another way to add/create object attribute inside the method </b></h3>

````python
class x:
    a = 100     # class attribute
    
    def m1(self):
        a = 120
        self.b = 200
        return 'm1 of class x', self.a, self.b, a
````

````python
class Car:
    tyres = 4
    color = 'red'
    
    def start(self):
        return "{} color car started".format(self.color)
    
    def stop(self):
        return '{} color car stopped'.format(self.color)
    
    def changeColor(self, color):
        self.color = color
        return 'Your car color changed to {}'.format(self.color)
````

````python
car1 = Car()
car2 = Car()
````

````python
car1.color
````

````python
car1.changeColor('Blue')
````

````python
car1.color
````

````python
car2.color
````
