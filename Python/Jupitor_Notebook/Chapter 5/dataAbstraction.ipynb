<h2><b><u>Abstraction in Python Coding</u></b></h2>
<p>Python Providing abstraction with the help of </p>
<ol>
    <li><b>Abstract Class   --></b> Make class as abstract class using ABC meta class</li>
    <li><b>Abstract Methods --></b> These methods must be implemented by child classes</li>
</ol>

<h3><b><u>Example 1</u></b></h3>

```python
from abc import ABC, abstractmethod
```

```python
type(ABC)
```

```python
type(abstractmethod)
```

```python
class Vehicle(ABC):
    a = 100
    b = 200
    
    @abstractmethod
    def start(self):
        pass
    
    @abstractmethod
    def stop(self):
        pass
```

```python
veh1 = Vehicle()
```

```python
class Car(Vehicle):
    def start(self):
        return 'car started'
```

```python
car1 = Car()
```

```python
class Vehicle(ABC):
    a = 100
    b = 200
    
    @abstractmethod
    def start(self):
        pass
    
    @abstractmethod
    def stop(self):
        pass


class Bike(Vehicle):
    tyres = 4
    def __init__(self, color='white'):
        self.color = color
        print(f'{self.color} color car created')
        
    def start(self):
        return 'Car started'
    
    def stop(self):
        return 'car stopped'
    
```

```python
bike1 = Bike('Red')
```

```python
bike1.start()
```

```python
bike1.stop() 
```

```python
bike1.tyres
```

```python
bike1.color
```

```python
bike1.start()
```

<hr><h5><b><i>
<li>If any class inherited with ABC meta class it becomes an abstract class</li>
<li>If any method created with @abstractmethod decorator method It must be impleted in the child class</li>
<li>If child class does not implement abstract method object will not be created because abstraction force rule </li><br>

<b><i>Why we are using abstraction?</i></b>
<ol>
    <li>Security</li>
        <ul>
            <li>Hide Internal Implementation</li>
        </ul>
    <li>Simplicity</li>
        <ul>
            <li>User see only required things</li>
        </ul>
    <li>Standard structure</li>
        <ul>
            <li>All child classes must follow same methods</li>
        </ul>
</ol>
</i></b></h5>

```python
from abc import ABC, abstractmethod

class Rupay(ABC):
    
    @abstractmethod
    def pay(self, amount):
        pass
    

class PhonePe(Rupay):
    def pay(self, amount):
        print(f'Amount Paid {amount} using PhonePe')
        
class GooglePay(Rupay):
    def pay(self, amount):
        print(f'Amount Paid {amount} using Google Pay')
```

```python
pay1 = PhonePe()
pay1.pay(500)
```

```python
pay2 = GooglePay()
pay2.pay(1000)
```

<h3><b><u>Payment of employee salary system</u></b></h3>

```python
from abc import ABC, abstractmethod

class Employee(ABC):
    @abstractmethod
    def calculate_salary(self):
        pass
    

class FullTimeEmployee(Employee):
    def calculate_salary(self, sal):
        TotalSal = sal + (sal*0.4) + (sal*0.3) + (sal*0.2)
        grossSal = TotalSal - (TotalSal*0.05)
        print(f'Salary for current month is {grossSal}')
        
class PartTimeEmployee(Employee):
    def calculate_salary(self, sal):
        grossSal = sal - 200
        print(f'Salary for current month is {grossSal}')
```

```python
emp1 = FullTimeEmployee()
emp1.calculate_salary(10000)
```

```python
emp2 = PartTimeEmployee()
emp2.calculate_salary(8000)
```

<h3><i><b>What is the differnece between abstraction and Encapsulation</b></i></h3>

| **Encapsulation** | **Abstraction** |
|-------------------|-----------------|
| Hides **data** | Hides **implementation** |
| Uses **private/protected members** | Uses **abstract classes and abstract methods** |
| Focuses on **security** | Focuses on **simplicity** |
| **Example:** Car's `topSpeed` data | **Example:** Car's braking system implementation |
