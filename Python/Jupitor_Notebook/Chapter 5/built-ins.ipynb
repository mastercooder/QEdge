<h3><b>13. bytes </b></h3>
<li> Bytes are used to represent binary data - raw data like files, images, pdf, network data </li>
<li>bytes function returns a byte object </li>
<li>byte object is immutable sequence of interger's from 0 to 255</li>

<li>Syntax: bytes(source, encoding, errors)</li>
<li>source ===> data to convert into bytes</li>
<li>encoding ===> required if source is a string</li>
<li>errors ===> optional error handling ('strick', ignore')</li>

````python
x = bytes(65)
````

````python
type(x)
````

````python
print(x)
````

````python
for i in x:
    print(i, end=" ")
````

<h3><b>16. callable </b></h3>
<i> It tell us that weather that variable, funtion etc. is Callable or not </i>

````python
a = 10
````

````python
def f1():
    return "welcome"
````

````python
a()
````

````python
f1()
````

````python
callable(a)
````

````python
callable(f1)
````

````python
if callable(f1):
    print(f1())
````

````python
if callable(a):
    print(a())
else:
    print("It is not a Callable Object")
````

````python
obj = input("")

if callable(obj):
    print(obj())
else:
    print("It is not a Callable Object")
````
