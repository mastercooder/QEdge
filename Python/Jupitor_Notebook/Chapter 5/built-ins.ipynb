<h3><b>2. ord </b></h3>
<i>It Returns ascii number of the given character</i>

```python
type(ord)
```

```python
print(" Ascii Values")
print("---------------")

print('A:     ', ord('A'))
print('@:     ', ord('@'))
print('Enter: ', ord('\n'))
print('Tab:   ', ord('\t'))
```

<h3><b>3. chr </b></h3>
<i>It returns character for given ascii number</i>

```python
print(" chr Values")
print("------------")

print('65: ', chr(65))
print('9:  ', chr(9))
print('10: ', chr(10))
print('36: ', chr(36))
```

```python
s = "\\u0924\\u0947\\u091c\\u0947\\u0936 \\u092a\\u091f\\u0947\\u0932 \\u092a\\u093e\\u092f\\u0925\\u0928 \\u0938\\u0940\\u0916 \\u0930\\u0939\\u093e \\u0939\\u0948"

print(s.encode().decode('unicode_escape'))
```
