<h3><b><u>Types of Access In Python</u></b></h3>

| Access Modifier | Variable Naming | Description |
|-----------------|-----------------|-------------|
| **Public** | `variableName` | Can be accessed from anywhere by any object. |
| **Protected** | `_variableName` | Should not be accessed directly from outside the class (intended for internal use and subclasses). |
| **Private** | `__variableName` | Strongly hidden using name mangling; intended to be accessed only within the class. |


```python
class Student:
    def __init__(self, name, ):
        self.name = name #public variable
        
student1 = Student('Tejesh')
```

```python
student1.name
```

```python
class x:
    a = 100    # Public
    _b = 200   # Protected
    __c = 300  # Private
    
    def m1(self):
        return 'm1 of class x'
```

```python
print(x.a)
print(x._b)
```

```python
# IT is a private variable can't access it direcly
print(x.__c)
```

```python
xobj1 = x()
```

```python
xobj1.a
```

```python
xobj1._b
```

```python
xobj1.__C
```

```python
class x:
    a = 100    # Public
    _b = 200   # Protected
    __c = 300  # Private
    
    def m1(self):
        return 'm1 of class x'
    
    def displayC(self):
        return x.__c
```

```python
xobj1 = x()
```

```python
xobj1.displayC()
```

```python
# We need to class name to call the private value
x._x__c
```

```python
xobj1._x__c
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self._salary = salary              # Protected Valribale
        
    def set_salary(self, amount):
        if amount > 15000:
            self._salary = amount
        else:
            print('Invalid Salary')
            
    def get_salary(self):
        return self._salary
```

```python
emp1 = Employee('Giri', 10000)
```

```python
emp1._salary
```

```python
emp1.set_salary(3500)
```

```python
emp1.set_salary(23000)
```

```python
emp1._salary
```

```python
emp1.get_salary()
```
