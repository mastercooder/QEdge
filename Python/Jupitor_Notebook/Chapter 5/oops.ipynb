<h1 align='center'><b><u>Object-Oriented Programming System</u></b></h1>
<hr>

<i><b>It is a programming style where we organize entire code using
<ol>
    <li>Classes</li>
    <li>objects</li>
</ol><br>
Python support or implements OOPS completely
</b></i>


<hr>
