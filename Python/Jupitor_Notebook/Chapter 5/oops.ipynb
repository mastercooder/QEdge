<h3><b>i. <u>Single Inheritance</u></b></h3>

```python
# This Class will be called: Parent / Super / Base 
class c1:
    a = 100
    def m1(self):
        return 'm1 of class c1'
    
    
# Child / sub / derived
class c2(c1):
    b = 200
    def m2(self):
        return 'm2 of class c2'
```

```python
c2obj = c2()
```

```python
print(c2obj.a)
print(c2obj.b)
print(c2obj.m1())
print(c2obj.m2())
```

<h3><b>ii. <u>Multiline Inheritance</u></b></h3>

```python
class c3:
    a = 100
    def m3(self):
        return 'm3 of class c3'
    

class c4(c3):
    b = 200
    def m4(self):
        return 'm4 of class c4'
    

class c5(c4):
    c = 300
    def m5(self):
        return 'm5 of class c5'
```

```python
c5obj1 = c5()
```

```python
print(c5obj1.a)
print(c5obj1.b)
print(c5obj1.c)
print(c5obj1.m3())
print(c5obj1.m4())
print(c5obj1.m5())
```

<h3><b>iii. <u>Multiple Inheritance</u></b></h3>

```python
class c6:
    a = 100
    def m6(self):
        return 'm6 of class c6'
    
    
class c7:
    b = 200
    def m7(self):
        return 'm7 of class c7'
    

# This is called Multiple Inheritance   
# Getting data from 2 Individual Parents
class c8(c6, c7):
    c = 300
    def m8(self):
        return 'm8 of class c8'
```

```python
c8obj1 = c8()
```

```python
print(c8obj1.a)
print(c8obj1.b)
print(c8obj1.c)
print(c8obj1.m6())
print(c8obj1.m7())
print(c8obj1.m8())
```

<h3><b>iv. <u>Hirarical Inheritance</u></b></h3>

```python
class c9:
    a = 100
    def m9(self):
        return 'm9 of class c9'
    

class c10(c9):
    b = 200
    def m10(self):
        return 'm10 of class c10'
    

class c11(c9):
    c = 300
    def m11(self):
        return 'm11 of class c11'
```

```python
c10obj1 = c10()
```

```python
print(c10obj1.a)
print(c10obj1.b)
print(c10obj1.m9())
print(c10obj1.m10())
```

```python
c11obj1 = c11()
```

```python
print(c11obj1.a)
print(c11obj1.c)
print(c11obj1.m9())
print(c11obj1.m11())
```

<h3><b>v. <u>Hybrid Inheritance</u></b></h3>

```python
class c12:
    a = 100
    def m12(self):
        return 'm12 of class m12'
    
    
    class c13:
        b = 200
        def m13(self):
            return 'm13 of class c13'
```
