### 6. Write a funtion to find given number is prime or not
- untion name ==> is_prime(7)

```python
def is_prime(user):
    for i in range(2, user):
        if(user % i == 0):
            return "{} is Not a Prime".format(user)
            break
        
        else:
            return "{} is a Prime Number".format(user)
            break
 

try:
    user = int(input("Enter any Number: "))
    result = is_prime(user)
    print(result)
except Exception:
    print("Enter a Valid Input")
```

### 12. password validation
- Password rules
    - minimum 8 characters
    - 1 upper case 
    - 1 lower case
    - 1 digit 
    - 1 special character

```python
import re

userInput = input("Enter a Password: ")

pattern = r"^(?=.*[A-Z])(?=.*[a-z])(?=.*[0-9])(?=.*[^A-Za-z0-9]).{8,13}$"

if re.fullmatch(pattern, userInput):
    print("Password Accepted")
else:
    print("Please Chose a Hard Password")
```

### 13. Develop mini project 
- create user Registration from validator using
    - funtions 
    - exceptions
    - regex
    
- validate
    - name
    - email
    - phone
    - password

```python
import re

def validate_name(name):
    if re.fullmatch(r'[A-Za-z]{4,20}', name):
        return True
    raise ValueError("Invalid Name!")

def validate_email(email):
    if re.fullmatch(r'[a-zA-Z0-9._]+@[a-zA-Z]+\.[a-zA-Z]{2,}', email):
        return True
    raise ValueError("Invalid Email!")

def validate_phone(phone):
    if re.fullmatch(r'[6-9][0-9]{9}', phone):
        return True
    raise ValueError("Invalid Phone No!")

def validate_password(password):
    if re.fullmatch(r'^(?=.*[A-Z])(?=.*[a-z])(?=.*[^A-Za-z0-9]).{8,}\S$', password):
        return True
    
    raise ValueError(
        "Invalid Password! Must contain:\n"
        "- At least 8 characters\n"
        "- One uppercase letter\n"
        "- One lowercase letter\n"
        "- One digit\n"
        "- One special character"
    )


def register_user():
    try:
        userName = input("Enter the Name: ")
        validate_name(userName)
        
        userEmail = input("Enter the Email: ")
        validate_email(userEmail)
        
        userPhone = input("Enter the Phone No: ")
        validate_phone(userPhone)
        
        userPassword = input("Enter the Password: ")
        validate_password(userPassword)
        
        print("\n===== REGISTRATION SUCCESSFUL =====")
        print("Name :", userName)
        print("Email:", userEmail)
        print("Phone:", userPhone)
        print("Password: ", userPassword)
        
    except ValueError as e:
        print("Registration Error")
        print(e)
        

register_user()
```

### 17. Count Vowels from given string text 

```python
import re

searchStaring = "count Vowels from given string text"

pattern = 'a|e|i|o|u'
count = 0


for vovle in searchStaring:
    if re.search(pattern, vovle, re.I):
        count+=1
    
print("The No of Vovles in the String is {}".format(count))
```
