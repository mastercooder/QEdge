<h2><b>1. <u>Timer</u></b></h2>


```python
import time

def t1(s):
    mins, secs = divmod(s, 60)
    print("Loading... {} mins and {} seconds".format(mins, secs))

    while s:
        mins, secs = divmod(s, 60)
        print(f"{mins:02d}:{secs:02d}")
        time.sleep(1)
        s -= 1

    print("Done")
```

```python
t1(10) 
```
