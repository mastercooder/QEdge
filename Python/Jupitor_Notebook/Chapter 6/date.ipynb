<h2><b><u>Date Formats</u></b></h2>
<ol>
    <li>%Y ==> Four-digit Year( 2025)</li>
    <li>%m ==> month as a zero-padded decimal number(01-12)</li>
    <li>%d ==> Day of the month as a zero-padding decimal number(01-31)</li>
    <li>%H ==> Hour( 24 hours clock)</li>
    <li>%M ==> Minues (01-60)</li>
    <li>%S ==> seconds(01-60)</li>
    <li>%A ==> Week day as full name (Sunday - Saturday)</li>
    <li>%a ==> Week day as short name (sun - sat)</li>
    <li>%B ==> Month name as full name( January - December )</li>
    <li>%b ==> Month name as short name ( Jan - Dec )</li>
</ol>
