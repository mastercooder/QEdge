```python
customer_contacts = [
'567-456-5823',
'567-436-5823',
'567-446-5823',
'567-456-9823',
'567-456-7823',
'56k-456-2823'
]

for i in customer_contacts:
    print(i)
    
    
import re

for i in customer_contacts:
    contactGrp = re.search("(?P<State>[1-9]{3})-(?P<Exchange>[0-9]{3})-(?P<Phone>[0-9]{4})", i)
    if contactGrp and contactGrp.groupdict()['State'] == '567':
        print(contactGrp)
```
