```python
import re

courses = {
    'python': 12000,
    'react': 4500,      
    'django': 7000,
    'fastapi': 5500,
    'java': 8000,
    'javascript': 6000
}

s_course = input("Enter Course Name: ").replace(" ", "")

fileName = 'txt files/students.txt'

for i in open(fileName, 'r'):
    if re.fullmatch(re.escape(s_course), i.split(",")[2].replace(" ", "")):
        sid = i.split(",")[0]
        s_name = i.split(",")[1].replace(" ", "").capitalize()
        s_fee = int(i.split(",")[3])
        due_fee = courses[s_course.lower()] - s_fee
        print("{} Due Fee is {}".format(s_name, due_fee))
        
```
