    3. Ask Fee to pay for given course
    4. It has to generate a new Student ID

```python
import re

courses = {
    'python': 12000,
    'react': 4500,      
    'django': 7000,
    'fastapi': 5500,
    'java': 8000,
    'javascript': 6000
}

s_course = input("Enter Course Name: ").replace(" ", "")

fileName = 'txt files/students.txt'

for i in open(fileName, 'r'):
    if re.fullmatch(re.escape(s_course), i.split(",")[2].replace(" ", "")):
        sid = i.split(",")[0]
        s_name = i.split(",")[1].replace(" ", "").capitalize()
        s_fee = int(i.split(",")[3])
        due_fee = courses[s_course.lower()] - s_fee
        print("{} Due Fee is {}".format(s_name, due_fee))
        
```

```python
import re

sid = input("Enter Student SID: ").replace(" ", "")

for i in open('txt files/students.txt', 'r'):
    if re.fullmatch(re.escape(sid), i.split(",")[0].replace(" ", "")):
        s_name = i.split(",")[1].replace(" ", "").capitalize()
        s_course = i.split(",")[2].replace(" ", "").capitalize()
        paid_fee = int(i.split(",")[3])
        
    for j in open('txt files/courses.txt', 'r'):
        if re.fullmatch(re.escape(s_course), j.split(",")[0].replace(" ", "")):
            total_fee = j.split(",")[1]
  
    
due_fee = int(total_fee) - paid_fee
print("The SID:{} Represents {} || Fee Paid: {} || Fee Pending: {}".format(sid, s_name, paid_fee, due_fee))
```
