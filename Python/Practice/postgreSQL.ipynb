```python
import psycopg2

conn = psycopg2.connect(
    dbname = 'postgres',
    user = 'postgres',
    password = '14422',
    host = 'localhost'
)

cur = conn.cursor()
```

```python
cur.execute("""
CREATE TABLE students(
    sid INTEGER,
    sname TEXT,
    scourse TEXT,
    fee INTEGER
)
""")

conn.commit()

print("Table Created Successfully!")
```
