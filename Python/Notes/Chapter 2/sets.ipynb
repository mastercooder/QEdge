```python
#sets
#unordered colllection of unique elements
#used to store multipllems without duplicates
#It performs mathematical set operations
#it has union , intersect, minus ...
#sets doesn't have any indexing or slicing
```

```python
s1 = {10,20,30,40,50}
s2 = {40,50,60,70,80}
```

```python
print(s1)
print(s2)
print(s1 - s2)  # s1  only
print(s2 - s1)  # s2 only
print(s1|s2)  # s1 union s2
print(s1&s2)  # s1 intersection s2
print(s1^s2)  # s1 symmetric_difference s2
```
