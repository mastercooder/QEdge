```python
i=5
while i>=1:
    j=i
    while j>=1:
        print(j,end="")
        j = j - 1
    print()
    i=i-1
```

```python
i=1
n=5
while i<=n:
    j = 1
    while j<=n-i:
        print(" ",end="")
        j+=1

    k = 1
    while k<=i:
        print("*",end="")
        k+=1

    print()
    i+=1
```

```python
i=1
n=5
while i<=n:
    j = 1
    while j<=n-i:
        print(" ",end="")
        j+=1

    k = 1
    while k<=(2*i-1):
        print("*",end="")
        k+=1

    print()
    i+=1
```

```python
i=5
n=5
while i>=1:
    j = 1
    while j <= n-i:
        print(" ",end="")
        j += 1

    k = 1
    while k<=i:
        print("*",end="")
        k+=1

    print()
    i-=1
```

```python
#inverted pyramid
i=5
n=5
while i>=1:
    j = 1
    while j<=n-i:
        print(" ",end="")
        j+=1

    k = 1
    while k<=(2*i-1):
        print("+",end="")
        k+=1

    print()
    i-=1
```

```python
#AP  ==> Arithmetic Progression
start = 2
diff = 3
for i in range(30):
    print(start+i*diff,end=" ")
```

```python
#GP   ==> Geometric Progression
start = 2
ratio = 2
for i in range(10):
    print(start*(ratio**i),end=" ")
```
