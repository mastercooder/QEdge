```python
v1 = input("Enter a integer value for V1 = ")
v2 = input("Enter a integer value for V2 = ")
try:
    v3 = int(v1) + int(v2)
except Exception as err:
    print("Error ==> ",err.with_traceback)
else:
    print(v3)
finally:
    print("Thanks")
```

```python
v1 = input("Enter a integer value for V1 = ")
v2 = input("Enter a integer value for V2 = ")
try:
    v3 = int(v1) + int(v2)
except Exception as err:
    print("Error Message ==> ",err.args[0])
else:
    print(v3)
finally:
    print("Thanks")
```

```python
course = input("Enter course name = ")
if course in courses:
    fee = courses.get(course)
    print("Fee for your given course {} is {} ".format(course,fee))
else:
    raise CourseNotFoundError
```

```python
course = input("Enter course name = ")
try:
    if course in courses:
        fee = courses.get(course)
    else:
        raise CourseNotFoundError
except(CourseNotFoundError):
    print("Given course is not existed")
else:
    print("Fee for your given course {} is {} ".format(course,fee))
finally:
    print("Thanks")
```
