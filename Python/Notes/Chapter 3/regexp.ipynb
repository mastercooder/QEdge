```python
# \s  === only space
str1 = input("Enter a string value = ")
if re.search('^\s$',str1):
    print("Matched    ")
else:
    print("Not Matched")
```

```python
# \S  === only space
str1 = input("Enter a string value = ")
if re.search('^\S$',str1):
    print("Matched    ")
else:
    print("Not Matched")
```

```python
# \w  === 
str1 = input("Enter a string value = ")
if re.search('^\w$',str1):
    print("Matched    ")
else:
    print("Not Matched")
```
