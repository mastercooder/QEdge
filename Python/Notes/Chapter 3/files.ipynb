```python
#total number of students for given course
```

```python
fobj = open("data/students.txt")
course = input("Enter course Name = ")
count=0
for i in fobj:
    if re.search(course,i,re.I):
        count+=1
else:
    print("Total students joined for {} course is {}".format(course.upper(),count))
```

```python
#total revenue for given course
```

```python
"1001,ravi,python,3500".split(",")
```

```python
"1001,ravi,python,3500".split(",")[3]
```

```python
fobj = open("data/students.txt")
course = input("Enter course Name = ")
rev=0
for i in fobj:
    if re.search(course,i,re.I):
        rev = rev + int(i.split(",")[3])
else:
    if rev==0:
        print("No such course is existed")
    else:
        print("Total Revenue for {} course is {}".format(course.upper(),rev))
```
