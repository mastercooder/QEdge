```python
#Nested if
maths = int(input("Enter Marks for Maths = "))
physics = int(input("Enter Physics for Physics = "))
chemistry = int(input("Enter chemistry for chemistry = "))
maxMarks=100
minMarks=40
if maths<=maxMarks and physics<=maxMarks and chemistry<=maxMarks:
    if maths>=minMarks and physics>=minMarks and chemistry>=minMarks:
        print("Pass")
    else:
        print("Fail")
else:
    print("Invalid marks")
```

```python
maths = int(input("Enter Marks for Maths = "))
physics = int(input("Enter Physics for Physics = "))
chemistry = int(input("Enter chemistry for chemistry = "))
maxMarks=100
minMarks=40
if maths<=maxMarks and physics<=maxMarks and chemistry<=maxMarks:
    if maths>=minMarks:
        if physics>=minMarks:
            if chemistry>=minMarks:
                print("Pass")
            else:
                print("Failed in chemistry")
        else:
            print("Failed in physics")
    else:
        print("Failed in Maths")
else:
    print("Invalid marks")
```

```python
maths = int(input("Enter Marks for Maths = "))
physics = int(input("Enter Physics for Physics = "))
chemistry = int(input("Enter chemistry for chemistry = "))
maxMarks=100
minMarks=40
if maths<=maxMarks and physics<=maxMarks and chemistry<=maxMarks:
    if maths>=minMarks:
        if physics>=minMarks:
            if chemistry>=minMarks:
                print("*********************************")
                print("*******Status : Pass ************")
                print("*********************************")
                print("Maths       = ",maths)
                print("Maths       = ",physics)
                print("Maths       = ",chemistry)
                print("*********************************")
                print("Tatal Marks = ",maths+physics+chemistry)
                print("*********************************")
            else:
                print("*********************************")
                print("*******Status : Fail ************")
                print("*********************************")
                print("Maths       = ",maths)
                print("Maths       = ",physics)
                print("Maths       = ",chemistry)
                print("*********************************")
                print("Tatal Marks = ",maths+physics+chemistry)
                print("*********************************")
        else:
            print("*********************************")
            print("*******Status : Fail ************")
            print("*********************************")
            print("Maths       = ",maths)
            print("Maths       = ",physics)
            print("Maths       = ",chemistry)
            print("*********************************")
            print("Tatal Marks = ",maths+physics+chemistry)
            print("*********************************")
    else:
        print("*********************************")
        print("*******Status : Fail ************")
        print("*********************************")
        print("Maths       = ",maths)
        print("Maths       = ",physics)
        print("Maths       = ",chemistry)
        print("*********************************")
        print("Tatal Marks = ",maths+physics+chemistry)
        print("*********************************")
else:
    print("Invalid marks")
```
