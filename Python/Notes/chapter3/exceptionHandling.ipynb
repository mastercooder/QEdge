```python
#Exceptions
#try...except...else...finally....
#try -->  contains error code
#except   ---> will execute when there is error existed
#else     ---> if no error then this block will execute
#finally  ---> this block will execute always
#case1   ----> Error raised
#case2   ----> No error
#case3   ----> error is there but not mentioned in except block
```

```python
Exception Syntax
===================
try:
    error code 
except(er1,er2,er3,....,ern):
    exception code
else:
    code on no error
finally:
    execute always
```
