```python
def pymax(*args):
    if len(args)==0: return None
    mx = args[0]
    for i in args:
        if i>mx:
            mx=i
    return mx
```

```python
pymax(1,2,3,4,5)
```

```python
pymax(17,2,-3,-4,-5)
```

```python
def pymax2(**kwargs):
    mx = list(kwargs.values())[0]
    for i in kwargs.values():
        if i>mx:
            mx=i
    return mx
```

```python
pymax2(a=1,b=2,c=3,d=4)
```
