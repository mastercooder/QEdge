```python
#tuple   ===>  ( )
It is inbuilt class to store more one values with different data type objects
It is like list but read only object (no insert,no update,no delete)
It maintains duplicates
It is iterable object so that we can read with for loop
we can find out elements with indexing and slicing also
```

```python
#tuple creation
T1 = (10,20,30,40,50)
T2 = (10.45,10.65,10.75)
T3 = ('python','react','django')
T4 = (101,'python','ravi',10.45)
```

```python
type(T1)
```

```python
type(tuple)
```

```python
print(T1)
print(T2)
print(T3)
print(T4)
```
