## Data Structures

Data Structures are classified into two main categories:

```text
                         Data Structures
                                │
               ┌────────────────┴────────────────┐
               │                                 │
          In-built                        User-defined
```

### In-built Data Structures
1. List
2. Tuple
3. Dictionary
4. Sets

### User-defined Data Structures

#### Linear Data Structures
1. Array
2. Linked List
3. Stack
4. Queue

#### Non-Linear Data Structures
5. Tree
6. Graph
7. Hash Table

#### Python Collections (`collections` module)
1. Default Dictionary (`defaultdict`)
2. Ordered Dictionary (`OrderedDict`)
3. Deque (`deque`)
4. Named Tuple (`namedtuple`)
5. Counter (`Counter`)

---

## Hierarchical Representation

```text
Data Structures
│
├── In-built
│   ├── List
│   ├── Tuple
│   ├── Dictionary
│   └── Sets
│
└── User-defined
    ├── Linear
    │   ├── Array
    │   ├── Linked List
    │   ├── Stack
    │   └── Queue
    │
    ├── Non-Linear
    │   ├── Tree
    │   ├── Graph
    │   └── Hash Table
    │
    └── Python Collections
        ├── Default Dictionary (defaultdict)
        ├── Ordered Dictionary (OrderedDict)
        ├── Deque (deque)
        ├── Named Tuple (namedtuple)
        └── Counter (Counter)
```
