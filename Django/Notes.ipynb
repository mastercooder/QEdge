<h2><b><u>Python (.venv) Commands</u></b></h2> 

<h5><b><i>1. Create a Virtual Environment (.venv)*</i></b></h5> 

```bash
python -m venv .venv
```

<h5><b><i>2. Activate the Virtual Environment</i></b></h5> 

**Windows (CMD)**
```cmd
.venv\Scripts\activate
```

**Windows (PowerShell)**
```powershell
.venv\Scripts\Activate.ps1
```

<h5><b><i>3. Install Django 4.1</i></b></h5>

```bash
pip install django==4.1
```

<h5><b><i>4. Verify Django Installation</i></b></h5> 

```bash
python -m django --version
```
<br>
<h3><b>OR</b></h3>
<br>

```python
import django
print(django.get_version())
```

<h5><b><i>5. Deactivate the Virtual Environment</i></b></h5> 

```bash
deactivate
```

<h5><b><i>6. Create Another Virtual Environment with a Different Python Version</i></b></h5> 
<p><i><b>Note: The required Python version must already be installed.</b></i></p>

```bash
py -3.13 -m venv .venv313
```

or

```bash
py -3.11 -m venv .venv311
```

<h5><b><i>7. Activate the New Environment </i></b></h5>

```cmd
.venv313\Scripts\activate
```

<h5><b><i>8. Install Django 5.0 </i></b></h5> 

```bash
pip install django==5.0
```

<h5><b><i>9. Check Python Version </i></b></h5>  

```bash
python --version
```

<h5><b><i> 10. View Installed Packages</i></b></h5>  
```bash
pip list
```

<h5><b><i> 11. Freeze Dependencies</i></b></h5>  

```bash
pip freeze > requirements.txt
```

<h5><b><i>12. Install Dependencies from requirements.txt </i></b></h5> 

```bash
pip install -r requirements.txt
```

<h2><b><u>Steps</u></b></h2>

<h5><b><i>1. Create Enviroment with Python and Django</i></b></h5>

```bash
conda create --name env1 python==3.13
```

<h5><b><i>2. Activate Environment</i></b></h5>

```bash
conda activate env1
```

<h5><b><i>3. Go to the location where you want create django project</i></b></h5>

```cmd
cd location
```

<h5><b><i>4. Create .env</i></b></h5>

```bash
python -m venv .env
```

<h5><b><i>5. Activate Virtual Environment</i></b></h5>

```bash
.env\scripts\activate
```

<h5><b><i>6. Install Packages</i></b></h5>

```bash
pip install django==5.0
```

<h5><b><i>7. Create Project</i></b></h5>

```bash
django-admin startproject proj1
```

<h5><b><i>8. Go inside project Filies</i></b></h5>

```cmd
cd project_name
```

<h5><b><i>9. Open Project Inside Vs Code</i></b></h5>

```cmd
code .
```

<h5><b><i>10. Start Project</i></b></h5>

```bash
python manage.py runserver
```

<h5><b><i>11. Access Project in Browser</i></b></h5>

```cmd
copy url which is generated and paste on browser address bar
```

<h3><b><u>File Structure</u></b></h3>

```text
proj2/ ------------------------> Root Project Folder
│
├── .env/ ---------------------> Python Virtual Environment (Python 3.11)
│
└── mysite/ -------------------> Base Django Directory
    │
    ├── manage.py -------------> Start Development Server (Runs project tools)
    │
    ├── db.sqlite3 ------------> Database File
    │
    └── mysite/ ---------------> Inner Project Configuration Folder
        ├── __pycache__/
        ├── __init__.py
        ├── asgi.py
        ├── settings.py -------> Main Project Settings
        ├── urls.py -----------> URL Router / Mappings
        └── wsgi.py
```
